```python
import os
import numpy as np
from PIL import Image

input_folder = "../data/augmentation_test/"
output_folder = "cropped_spectrograms/"
os.makedirs(output_folder, exist_ok=True)

def random_crop(image, crop_ratio=0.5):
    width, height = image.size
    crop_width = int(width * crop_ratio)
    crop_height = int(height * crop_ratio)

    if crop_width == 0 or crop_height == 0:
        print("Fehler: Cropping zu klein!")
        return None

    x_start = np.random.randint(0, width - crop_width)
    y_start = np.random.randint(0, height - crop_height)

    return image.crop((x_start, y_start, x_start + crop_width, y_start + crop_height))

for filename in os.listdir(input_folder):
    if filename.endswith(".png") or filename.endswith(".jpg"):  
        img_path = os.path.join(input_folder, filename)
        img = Image.open(img_path)

        cropped_img = random_crop(img)
        
        if cropped_img is not None:
            output_path = os.path.join(output_folder, f"cropped_{filename}")
            cropped_img.save(output_path)
            print("Gespeichert:", output_path)

print("Fertig!")
```

```python
import os
import cv2
import numpy as np
from PIL import Image

input_folder = "../data/augmentation_test/"
output_folder = "time_stretched_spectrograms/"
os.makedirs(output_folder, exist_ok=True)

def time_stretch_image(image, stretch_factor=1.5):
    img_array = np.array(image)  # Bild zu NumPy-Array umwandeln
    height, width = img_array.shape[:2]

    # Neue Breite nach Stretchen berechnen
    new_width = int(width * stretch_factor)
    
    # Bild in der X-Achse (Zeitachse) stretchen
    stretched_img = cv2.resize(img_array, (new_width, height), interpolation=cv2.INTER_LINEAR)
    
    return Image.fromarray(stretched_img)  # Zurück in PIL-Format umwandeln

for filename in os.listdir(input_folder):
    if filename.endswith(".png") or filename.endswith(".jpg"):  
        img_path = os.path.join(input_folder, filename)
        img = Image.open(img_path)

        stretched_img = time_stretch_image(img, stretch_factor=1.5)  # Stretch um 1.5x
        
        if stretched_img is not None:
            output_path = os.path.join(output_folder, f"stretched_{filename}")
            stretched_img.save(output_path)
            print("Gespeichert:", output_path)

print("Fertig!")
```
